## 회귀

## 1. 단순 선형 회귀 (Simple Linear Regression)

- **문제 유형**: 회귀
- **목표**: `HouseAge` 1개 변수만 사용해서 주택 가격(`target`)을 예측
- **모델 특징**:
  - 입력 변수 1개와 타깃 변수 사이의 **직선 관계**를 학습하는 모델
  - 가장 기본적인 회귀 모델
  - 해석이 쉽고, 변수와 결과의 관계를 이해하기 좋음

**하나의 독립변수로 연속형 값을 예측하는 단순 선형 회귀 모델**

### 평가 지표
- **MSE (Mean Squared Error)**: 예측 오차를 제곱해 평균낸 값으로, **작을수록 좋음**
- **R² Score**: 모델 설명력을 나타내며, **1에 가까울수록 좋음**


```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.datasets import fetch_california_housing
import pandas as pd

# 캘리포니아 주택 가격 데이터셋 로드
data = fetch_california_housing()

# 독립 변수(X), 종속 변수(y) 설정
X = pd.DataFrame(data.data, columns=data.feature_names)  # 특성 데이터
y = data.target  # 목표 변수(중간 주택 가격)

# 단순 선형 회귀 형태를 확인하기 위해 하나의 특성만 선택
X_house_age = X[['HouseAge']]

# 데이터를 학습용(train) / 테스트용(test)으로 분리
X_train, X_test, y_train, y_test = train_test_split(
    X_house_age, y, test_size=0.3, random_state=42
)

# 선형 회귀 모델 생성
model = LinearRegression()

# 모델 학습
model.fit(X_train, y_train)

# 테스트 데이터에 대한 예측 수행
y_pred = model.predict(X_test)

# 회귀 성능 평가
mse = mean_squared_error(y_test, y_pred)  # 평균 제곱 오차
r2 = r2_score(y_test, y_pred)  # 결정 계수

# 결과 출력
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R² Score: {r2:.4f}")
```

## 2. 다중 선형 회귀 (Multiple Linear Regression)

- **문제 유형**: 회귀
- **목표**: 캘리포니아 주택 데이터의 여러 특성들을 사용해 주택 가격을 예측
- **모델 특징**:
  - 여러 개의 입력 변수(`MedInc`, `HouseAge`, `AveRooms` 등)를 함께 사용
  - 각 변수의 영향을 선형 결합 형태로 반영
  - 단순 선형 회귀보다 더 많은 정보를 활용할 수 있음

**여러 개의 독립변수로 연속형 값을 예측하는 다중 선형 회귀 모델**

### 평가 지표
- **MSE (Mean Squared Error)**: 예측 오차를 제곱해 평균낸 값으로, **작을수록 좋음**
- **R² Score**: 모델 설명력을 나타내며, **1에 가까울수록 좋음**

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.datasets import fetch_california_housing
import pandas as pd

# 캘리포니아 주택 가격 데이터셋 로드
data = fetch_california_housing()

# 전체 특성을 독립 변수로 사용
X = pd.DataFrame(data.data, columns=data.feature_names)  # 여러 개의 입력 변수
y = data.target  # 목표 변수(중간 주택 가격)

# 데이터를 학습용(train) / 테스트용(test)으로 분리
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 다중 선형 회귀 모델 생성
model = LinearRegression()

# 모델 학습
model.fit(X_train, y_train)

# 테스트 데이터 예측
y_pred = model.predict(X_test)

# 회귀 성능 평가
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# 결과 출력
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R² Score: {r2:.4f}")
```

## 3. 다항 회귀 (Polynomial Regression)

- **문제 유형**: 회귀
- **목표**: `AveRooms`와 주택 가격 사이의 **곡선 형태 관계**를 반영하여 예측
- **사용 모델**:
  - `PolynomialFeatures(degree=2)`로 입력값을 다항식 형태로 변환
  - 변환된 데이터를 **선형 회귀(Linear Regression)** 모델로 학습
- **모델 특징**:
  - 실제 학습 모델은 선형 회귀이지만,
  - 입력 데이터를 `X`, `X²` 형태로 확장해 **비선형 관계**를 표현할 수 있음

**선형 회귀를 기반으로 하지만, 다항 특성 변환을 통해 곡선 관계를 학습하는 다항 회귀 모델**

### 평가 지표
- **MSE (Mean Squared Error)**: 예측값과 실제값의 차이를 제곱한 평균으로, **작을수록 좋음**
- **R² Score**: 모델이 데이터를 얼마나 잘 설명하는지를 나타내며, **1에 가까울수록 좋음**

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.datasets import fetch_california_housing
import pandas as pd

# 캘리포니아 주택 가격 데이터셋 로드
data = fetch_california_housing()

# DataFrame으로 변환
df = pd.DataFrame(data.data, columns=data.feature_names)
df["MedHouseVal"] = data.target  # 목표 변수 추가

# 특정 특성만 선택
# 다항 회귀 확인을 위해 하나의 변수만 선택
# 여러 특성을 사용하면 X1, X2 → X1, X2, X1², X2², X1·X2와 같은 형태로 확장됨
X = df[["HouseAge"]]
y = df["MedHouseVal"]

# 데이터를 학습용(train) / 테스트용(test)으로 분리
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 다항 특성 변환기 생성 (2차 다항식)
poly = PolynomialFeatures(degree=2)

# 학습 데이터에 대해 다항 특성 변환 수행
X_train_poly = poly.fit_transform(X_train)

# 테스트 데이터도 동일한 기준으로 변환
X_test_poly = poly.transform(X_test)

# 선형 회귀 모델 생성
model = LinearRegression()

# 변환된 다항 특성으로 모델 학습
model.fit(X_train_poly, y_train)

# 테스트 데이터 예측
y_pred = model.predict(X_test_poly)

# 회귀 성능 평가
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# 결과 출력
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R² Score: {r2:.4f}")
```

## 4. 로지스틱 회귀 분류 (Logistic Regression)

- **문제 유형**: 분류
- **목표**: Iris 데이터셋의 꽃 품종(class)을 예측
- **모델 특징**:
  - 이름에 회귀가 들어가지만 **분류 문제**에 사용되는 모델
  - 각 클래스에 속할 확률을 계산한 뒤 가장 높은 클래스를 예측
  - 비교적 단순하고 빠르며, 기본 분류 모델로 많이 사용됨

**입력 특성을 바탕으로 꽃의 품종을 분류하는 로지스틱 회귀 분류 모델**

### 평가 지표
- **Accuracy**: 전체 데이터 중 정답을 맞춘 비율
- **Confusion Matrix**: 실제 클래스와 예측 클래스가 어떻게 다른지 표로 확인
- **Classification Report**: Precision, Recall, F1-score를 함께 제공

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, roc_auc_score
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Iris 데이터셋 로드
data = load_iris()
X = data.data  # 특성 데이터
y = data.target  # 목표 변수(꽃 품종 레이블)

# 데이터를 학습용(train) / 테스트용(test)으로 분리
# stratify=y를 사용하여 클래스 비율을 유지
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 스케일링 적용
# 로지스틱 회귀는 특성 스케일의 영향을 받을 수 있으므로 표준화 수행
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 로지스틱 회귀 모델 생성
model = LogisticRegression(max_iter=200)

# 모델 학습
model.fit(X_train_scaled, y_train)

# 클래스 예측
y_pred = model.predict(X_test_scaled)

# 분류 성능 평가
accuracy = accuracy_score(y_test, y_pred)  # 정확도
cm = confusion_matrix(y_test, y_pred)  # 오차 행렬
report = classification_report(y_test, y_pred, target_names=data.target_names)  # 정밀도/재현율/F1-score

# 결과 출력
print(f"Accuracy: {accuracy:.4f}")
print("\nConfusion Matrix:")
print(cm)
print("\nClassification Report:")
print(report)
```

## 5. 결정트리 분류 (Decision Tree Classifier)

- **문제 유형**: 분류
- **목표**: Iris 데이터셋의 꽃 품종(class)을 예측
- **모델 특징**:
  - 질문을 단계적으로 나누듯이 데이터를 분할하여 분류
  - 예를 들어 특정 특성이 어떤 값보다 큰지/작은지를 기준으로 가지를 나눔
  - 결과를 트리 구조로 표현할 수 있어 해석이 쉬운 편

**특성 값을 기준으로 데이터를 반복적으로 나누어 클래스를 예측하는 결정트리 분류 모델**

### 평가 지표
- **Accuracy**: 전체 데이터 중 정답을 맞춘 비율
- **Confusion Matrix**: 어떤 클래스를 어떤 클래스로 잘못 예측했는지 확인 가능
- **Classification Report**: Precision, Recall, F1-score를 한 번에 확인 가능

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, roc_auc_score
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Iris 데이터셋 로드
data = load_iris()
X = data.data  # 특성 데이터
y = data.target  # 목표 변수(꽃 품종 레이블)

# 데이터를 학습용(train) / 테스트용(test)으로 분리
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 결정트리 모델 생성
# 트리 기반 모델은 스케일링이 필수는 아님
model = DecisionTreeClassifier(random_state=42)

# 모델 학습
model.fit(X_train, y_train)

# 클래스 예측
y_pred = model.predict(X_test)

# 분류 성능 평가
accuracy = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=data.target_names)

# 결과 출력
print(f"Accuracy: {accuracy:.4f}")
print("\nConfusion Matrix:")
print(cm)
print("\nClassification Report:")
print(report)
```

## 6. K-최근접 이웃 분류 (K-Nearest Neighbors, KNN)

- **문제 유형**: 분류
- **목표**: Iris 데이터셋의 꽃 품종(class)을 예측
- **모델 특징**:
  - 새로운 데이터가 들어오면,
    주변의 가장 가까운 데이터 `K개`를 확인해 다수결로 분류
  - 여기서는 `n_neighbors=5` 이므로 가까운 5개 이웃을 참고
  - 학습 과정이 단순하지만, 예측 시 주변 데이터를 많이 참고함

**가까운 이웃들의 클래스를 기준으로 새로운 데이터를 분류하는 KNN 분류 모델**

### 평가 지표
- **Accuracy**: 전체 데이터 중 맞춘 비율
- **Confusion Matrix**: 클래스별 오분류 패턴 확인
- **Classification Report**: Precision, Recall, F1-score 제공

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, roc_auc_score
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

# Iris 데이터셋 로드
data = load_iris()
X = data.data  # 특성 데이터
y = data.target  # 목표 변수(꽃 품종 레이블)

# 데이터를 학습용(train) / 테스트용(test)으로 분리
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# 스케일링 적용
# KNN은 거리 기반 모델이므로 스케일 차이에 매우 민감함
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# KNN 모델 생성
model = KNeighborsClassifier(n_neighbors=5)

# 모델 학습
model.fit(X_train_scaled, y_train)

# 클래스 예측
y_pred = model.predict(X_test_scaled)

# 클래스별 확률 예측
y_proba = model.predict_proba(X_test_scaled)

# 분류 성능 평가
accuracy = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=data.target_names)
auc = roc_auc_score(y_test, y_proba, multi_class="ovr")

# 결과 출력
print(f"Accuracy: {accuracy:.4f}")
print("\nConfusion Matrix:")
print(cm)
print("\nClassification Report:")
print(report)
```

## 모델 비교 포인트

- **단순 선형 회귀**
  - 변수 1개만 사용
  - 가장 기본적인 회귀 모델
  - 해석이 쉬움
  - 변수 하나로 관계를 빠르게 파악할 때 사용

- **다중 선형 회귀**
  - 여러 변수를 함께 사용
  - 더 많은 정보를 반영 가능
  - 현실 데이터에 더 가깝게 적용 가능
  - 여러 요인이 동시에 영향을 줄 때 사용

- **다항 회귀**
  - 선형 회귀 기반이지만 비선형 관계를 표현 가능
  - 곡선 형태 패턴을 학습할 수 있음
  - 직선으로 설명이 안 되는 경우 사용

- **로지스틱 회귀**
  - 기본적인 분류 모델
  - 빠르고 단순하며 baseline으로 자주 사용됨
  - 분류 문제에서 첫 번째로 시도하는 모델

- **결정트리**
  - 해석이 쉬움
  - 분기 구조를 통해 예측 과정을 확인 가능
  - 과적합 위험이 있음
  - 규칙 기반으로 설명이 필요한 경우 사용

- **KNN**
  - 매우 직관적인 모델
  - 거리 계산 기반이므로 스케일링이 중요
  - 예측 시 계산량이 커질 수 있음
  - 데이터 간 유사도가 중요한 경우 사용
