```python
"""
실시간(준실시간) 날씨 Polling 예제
- OpenAPI(REST)로 현재 날씨를 주기적으로 조회
- Open-Meteo API 사용
- 핵심 포인트: "요청 -> 응답 -> 대기 -> 다시 요청" (Polling 구조)
"""

import time
import requests
```

```python
def fetch_weather(lat: float, lon: float) -> dict:
    """현재 시점의 날씨를 OpenAPI로 조회"""
    url = "https://api.open-meteo.com/v1/forecast"
    params = {
        "latitude": lat,
        "longitude": lon,
        "current": "temperature_2m",
    }

    resp = requests.get(url, params=params, timeout=10)
    resp.raise_for_status()
    return resp.json()["current"]
```

```python
def poll_weather(lat: float, lon: float, interval_sec: int = 60):
    """interval_sec마다 날씨를 반복 조회(Polling)"""
    while True:
        try:
            current = fetch_weather(lat, lon)

            print(
                f"[{current['time']}] "
                f"temp={current['temperature_2m']}°C "
            )

        except Exception as e:
            print("API 호출 실패:", e)

        time.sleep(interval_sec)
```

```python
LAT = 36.3553
LON = 127.2983

# 30초마다 현재 날씨 조회
poll_weather(LAT, LON, interval_sec=30)
```
