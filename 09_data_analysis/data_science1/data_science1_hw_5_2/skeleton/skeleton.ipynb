```python
import requests
from bs4 import BeautifulSoup
import json

# 1. 대상 URL 설정
url = "https://ko.wikipedia.org/wiki/자료"
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}
```

```python
# 2. 웹페이지 내용 가져오기
response = requests.get(url, headers=headers)
response.raise_for_status()
```

```python
# 3. HTML 파싱
soup = BeautifulSoup(response.text, 'html.parser')

# 3-1. find()로 본문 영역 찾기
content_div = soup.find('div', class_='mw-parser-output')
```

```python
# 4. find_all()로 문단(p) 전부 찾기
paragraphs = content_div.find_all('p')
paragraph_list = []

for para in paragraphs:
    # 4-1. 문단 텍스트 추출
    text = para.get_text().strip()
    if text:  # 내용이 비어있지 않은 문단만 추가
        paragraph_list.append(text)
```

```python
# 5. 데이터를 딕셔너리 형태로 정리 (content가 리스트가 됨)
data_to_save = {
    "title": "자료",
    "url": url,
    "content": paragraph_list  # [ "첫 번째 문단", "두 번째 문단", ... ]
}
```

```python
# 6. JSON 파일로 저장
with open("result.json", "w", encoding="utf-8") as f:
    json.dump(data_to_save, f, ensure_ascii=False, indent=4)

print("--- 'result.json' 파일에 문단 단위로 저장되었습니다 ---")
```
