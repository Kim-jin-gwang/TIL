이 실습에서는 scikit-learn에서 제공하는 Breast Cancer Wisconsin 데이터셋을 사용합니다.

- 목적: 종양이 '악성(malignant)'인지 '양성(benign)'인지 분류하는 이진 분류 문제
- 입력 데이터(X): 종양의 크기/형태/질감 등과 관련된 여러 개의 수치형 특성(feature)
- 타깃(y):
    - 0: 악성(malignant)
    - 1: 양성(benign)
- 특징:
  - 모든 feature가 수치형이므로 전처리 부담이 낮음
  - feature들의 스케일(값의 범위)이 서로 달라 KNN에서 표준화(StandardScaler)의 필요성을 학습하기에 적합
  - KNN은 거리(distance)를 기반으로 분류하므로 스케일링이 성능에 큰 영향을 줄 수 있음

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
# 1. 데이터 로드
# scikit-learn 내장 데이터셋을 로드합니다.
data = load_breast_cancer()   # TODO: 데이터셋 로드

# pandas DataFrame/Series로 변환하여 다루기 쉽게 만듭니다.
# feature는 DataFrame, target은 Series 형태로 구성합니다.
df = pd.DataFrame(data.data, columns=data.feature_names)     # TODO: feature 데이터로 DataFrame 생성
X = df      # TODO: 독립 변수(X) 설정
y = pd.Series(data.target)      # TODO: 종속 변수(y) 설정 (타깃)
```

```python
print(y)
```

```python
# 2. 학습 데이터와 테스트 데이터 분리
# train_test_split()을 사용해 데이터를 훈련 세트(70%)와 테스트 세트(30%)로 분리합니다.
# stratify=y는 종속 변수(y)의 비율을 유지한 상태로 데이터를 나누기 위한 옵션입니다. 이를 적용하세요.
X_train, X_test, y_train, y_test = train_test_split(
    X,y,test_size=0.3,stratify=y,random_state=42      # TODO

)
```

```python
# 3. 데이터 스케일링 (표준화)
# StandardScaler()를 사용해 평균 0, 표준편차 1로 표준화합니다.
# 훈련 데이터로만 스케일러 기준을 학습(fit)하고, 테스트 데이터는 transform만 해야 합니다.
scaler = StandardScaler()  # TODO: StandardScaler 생성
X_train_scaled = scaler.fit_transform(X_train)  # TODO: 훈련 데이터 표준화 (fit 포함)
X_test_scaled = scaler.transform(X_test)   # TODO: 테스트 데이터 표준화 (transform만)
```

```python
# 4. KNN 분류 모델 생성 및 학습
# n_neighbors=5로 설정하여 가장 가까운 5개 이웃으로 분류합니다.
knn = KNeighborsClassifier()    # TODO: KNeighborsClassifier 생성
knn.fit(X_train_scaled, y_train)  # TODO: 표준화된 훈련 데이터로 학습
```

```python
# 5. 테스트 셋에서 첫 번째 샘플 선택
# 첫 번째 샘플을 선택하여 일관된 결과를 보장합니다.
selected_sample_scaled = X_test_scaled[0:1]  # TODO: 첫 샘플을 (1, -1) 형태 2차원으로 만들기
```

```python
# 6. 실제 라벨과 예측 라벨 비교
actual_label = y_test.iloc[0]  # TODO: y_test에서 첫 샘플의 실제 라벨
predicted_label = knn.predict(selected_sample_scaled)[0]  # TODO: knn으로 첫 샘플 예측 라벨
```

```python
# 7. 결과 출력
print(f"실제 라벨: {'양성(benign)' if actual_label == 1 else '악성(malignant)'}")
print(f"예측 라벨: {'양성(benign)' if predicted_label == 1 else '악성(malignant)'}")
```
