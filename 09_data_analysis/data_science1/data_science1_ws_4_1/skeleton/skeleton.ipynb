```python
import requests
import pandas as pd
import re
from html import unescape
```

```python
# 네이버 OpenAPI 인증 정보
CLIENT_ID = "EDZj4GtOZnrUuOC5quUq"
CLIENT_SECRET = "EV09c2U784"

# 네이버 뉴스 검색 OpenAPI 엔드포인트
url = "https://openapi.naver.com/v1/search/news.json"

# 문제 1) 인증 정보를 포함한 headers 딕셔너리를 완성하시오.
headers = {
    "X-Naver-Client-Id": CLIENT_ID,
    "X-Naver-Client-Secret": CLIENT_SECRET
}
```

```python
# 문제 2) 검색 조건 파라미터를 완성하시오.
params = {
    "query": "인공지능",
    "display": 5,
    "start": 1
}
```

```python
# 문제 3) GET 요청을 보내고 응답을 받으시오.
response = requests.get(url, headers=headers, params=params)

# 문제 4) HTTP 상태 코드를 출력하고, 200이 아니면 에러를 발생시키시오.
print("status code:", response.status_code)
if response.status_code != 200:
    raise RuntimeError("API 호출 실패")
```

```python
# 문제 5) JSON 응답을 파싱하시오.
data = response.json()

# 문제 6) 뉴스 데이터가 들어 있는 items 리스트를 추출하시오.
items = data['items']
```

```python
# 뉴스 제목/요약에 포함된 HTML 태그 제거용 함수
def clean_html(text):
    if text is None:
        return ""
    text = unescape(text)
    return re.sub(r"<.*?>", "", text)
```

```python
# 문제 7) items에서 필요한 필드만 추출하여 rows 리스트를 완성하시오.
rows = []
for it in items:
    rows.append({
        "title": it.get('title'),
        "description": it.get('description'),
        "link": it.get('link'),
        "pubDate": it.get('pubDate')
    })
```

```python
# 문제 8) rows를 DataFrame으로 변환하시오.
df = pd.DataFrame(rows)

# 결과 확인
print(df.head())
```
