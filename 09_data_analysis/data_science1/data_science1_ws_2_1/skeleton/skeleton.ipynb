```python
import pandas as pd

df = pd.read_csv("../data/ab_user_level_final.csv")

```

```python
# TODO:
# 사용자 단위 CTR을 계산하세요.
# click 값을 impression으로 나누어 CTR 컬럼을 생성합니다.
df["CTR"] = df['click'] / df['impression']
```

```python
# TODO:
# 사용자 단위 Conversion을 계산하세요.
# purchase 값을 impression으로 나누어 Conversion 컬럼을 생성합니다.
df["Conversion"] = df['purchase'] / df['impression']
```

```python
# TODO:
# 사용자 단위 Revenue_per_user 컬럼을 생성하세요.
# 현재 데이터는 이미 user-level이므로 revenue 값을 그대로 사용합니다.
df["Revenue_per_user"] = df['revenue']
```

```python
# TODO:
# variant, country, device 기준으로 그룹화하고
# CTR, Conversion, Revenue_per_user의 평균을 계산하세요.
grouped = df.groupby(
    ['variant', 'country', 'device']
).agg({
    "CTR": "mean",
    "Conversion": "mean",
    "Revenue_per_user": "mean"
}).reset_index()

print(grouped.head())
```

```python
# TODO:
# 전체 CTR, 전체 Conversion, 전체 Revenue 평균을 계산하여 출력하세요.
print("전체 CTR:", df['CTR'].mean())
print("전체 Conversion:", df['Conversion'].mean())
print("전체 Revenue:", df['Revenue_per_user'].mean())
```
