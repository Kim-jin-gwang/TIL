```python
import pandas as pd
```

```python
df = pd.read_csv("../data/ab_user_level_final.csv")
```

```python
# TODO:
# Conversion 컬럼을 생성하세요.
# purchase가 1번 이상이면 1, 아니면 0
df["converted"] = (df["purchase"] > 0).astype(int)
```

```python
# TODO:
# Conversion 컬럼을 생성하세요.
# purchase가 1번 이상이면 1, 아니면 0
df["converted"] = (df["purchase"] > 0).astype(int)
```

```python
print("\n=== Country 기준 분석 ===")
# TODO:
# country와 variant 기준으로 그룹화한 뒤
# Conversion과 CTR 평균을 계산하세요.
country_result = df.groupby(['country', 'variant']).agg({
    "converted" : "mean",
    "click": lambda x : (x>0).mean()
})

print(country_result)
```

```python
print("\n=== Device 기준 분석 ===")

# TODO:
# device와 variant 기준으로 그룹화한 뒤
# Conversion과 CTR 평균을 계산하세요.
device_result = df.groupby(['device', 'variant']).agg({
    "converted" : "mean",
    "click": lambda x : (x>0).mean()
})

print(device_result)
```
