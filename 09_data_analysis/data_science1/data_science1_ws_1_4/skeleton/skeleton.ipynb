```python
import pandas as pd
import numpy as np
from scipy.stats import norm
```

```python
df = pd.read_csv("../data/ab_user_level_final.csv")
```

```python
group_A = df[df["variant"] == 0]
group_B = df[df["variant"] == 1]

n_A = len(group_A)
n_B = len(group_B)
```

```python
# Conversion
# TODO:
# 각 그룹의 Conversion rate를 계산하세요.
# purchase가 1번 이상 발생한 사용자 비율을 평균으로 구하면 됩니다.
conv_A = (group_A['purchase'] > 0).mean()
conv_B = (group_B['purchase'] > 0).mean()
```

```python
# TODO:
# 각 그룹에서 purchase가 1번 이상 발생한 사용자 수를 계산하세요.
success_A = (group_A['purchase'] > 0).sum()
success_B = (group_B['purchase'] > 0).sum()
```

```python
# TODO:
# 두 그룹을 합친 전체 기준 성공 비율을 계산하세요.
# 식: (A 성공 수 + B 성공 수) / (A 전체 수 + B 전체 수)
# 이 값은 두 그룹이 원래 차이가 없다고 가정할 때 사용하는 전체 기준 비율입니다.
p_pool = (success_A + success_B) / (len(group_A) + len(group_B))
```

```python
# TODO:
# Conversion 검정을 위한 표준오차(SE)를 계산하세요.
SE = np.sqrt(p_pool * (1-p_pool) * (1/len(group_A) + 1/len(group_B)))

# TODO:
# Z-score를 계산하세요.
Z = (conv_B - conv_A) / SE

# TODO:
# 양측 검정 p-value를 계산하세요.
p_value = (1 - norm.cdf(abs(Z))) * 2

print("=== Conversion Test ===")
print("conv_A:", conv_A)
print("conv_B:", conv_B)
print("Z:", Z)
print("p-value:", p_value)
```

```python
if p_value < 0.05:
    if conv_B > conv_A:
        print("해석: B 그룹의 Conversion이 A 그룹보다 유의하게 높습니다.")
    else:
        print("해석: A 그룹의 Conversion이 B 그룹보다 유의하게 높습니다.")
else:
    print("해석: Conversion 차이는 통계적으로 유의하지 않습니다.")
```

```python
# CTR
# TODO:
# 각 그룹의 CTR을 계산하세요.
# click이 1번 이상 발생한 사용자 비율을 평균으로 구하면 됩니다.
ctr_A = (group_A['click'] > 0).mean()
ctr_B = (group_B['click'] > 0).mean()

# TODO:
# 각 그룹에서 click이 1번 이상 발생한 사용자 수를 계산하세요.
click_A = (group_A['click'] > 0).sum()
click_B = (group_B['click'] > 0).sum()
```

```python
# TODO:
# CTR에 대한 전체 기준 성공 비율을 계산하세요.
p_pool_ctr = (ctr_A + ctr_B) / (len(group_A) + len(group_B))

# TODO:
# CTR 검정을 위한 표준오차(SE)를 계산하세요.
SE_ctr = np.sqrt(p_pool_ctr * (1-p_pool_ctr) * 1/len(group_A) * 1/len(group_B))

# TODO:
# CTR의 Z-score를 계산하세요.
Z_ctr = (ctr_B - ctr_A) / SE_ctr

# TODO:
# CTR의 양측 검정 p-value를 계산하세요.
p_value_ctr = (1 - norm.cdf(abs(Z_ctr))) * 2

print("\n=== CTR Test ===")
print("ctr_A:", ctr_A)
print("ctr_B:", ctr_B)
print("Z:", Z_ctr)
print("p-value:", p_value_ctr)
```

```python
if p_value_ctr < 0.05:
    if ctr_B > ctr_A:
        print("해석: B 그룹의 CTR이 A 그룹보다 유의하게 높습니다.")
    else:
        print("해석: A 그룹의 CTR이 B 그룹보다 유의하게 높습니다.")
else:
    print("해석: CTR 차이는 통계적으로 유의하지 않습니다.")
```

```python
print("\n=== 종합 판단 ===")
if p_value < 0.05 and conv_B > conv_A:
    print("결론: Conversion 기준으로 B안을 검토할 수 있습니다.")
elif p_value < 0.05 and conv_A > conv_B:
    print("결론: Conversion 기준으로 기존 A안을 유지하는 것이 적절합니다.")
else:
    print("결론: Conversion 기준으로 추가 실험 또는 데이터 수집이 필요합니다.")

if p_value_ctr < 0.05:
    print("참고: CTR도 유의한 차이가 있어 클릭 단계에서도 변화가 발생했습니다.")
else:
    print("참고: CTR은 유의하지 않아 클릭 단계 변화는 제한적으로 해석해야 합니다.")
```
