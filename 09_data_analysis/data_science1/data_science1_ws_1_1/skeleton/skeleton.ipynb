```python
import pandas as pd
```

```python
# TODO:
# 세 개의 CSV 파일을 각각 읽어 user_df, exposure_df, action_df에 저장하세요.
# 사용자 기본 정보, 노출 로그, 행동 로그를 분리해서 불러와야 이후 분석 단계에서
# 각 데이터의 역할을 구분하여 사용할 수 있습니다.
user_df = pd.read_csv("../data/ab_user_master_2000.csv")
exposure_df = pd.read_csv("../data/ab_exposure_log_2000.csv")
action_df = pd.read_csv("../data/ab_action_log_2000.csv")
```

```python
# 각 데이터프레임의 shape, columns, head()를 출력합니다.
# shape는 행과 열의 크기를, columns는 컬럼 구성을,
# head()는 실제 데이터 샘플을 확인하는 데 사용합니다.
print(user_df.shape)
print(user_df.columns)
print(user_df.head())

print(exposure_df.shape)
print(exposure_df.columns)
print(exposure_df.head())

print(action_df.shape)
print(action_df.columns)
print(action_df.head())
```

```python
# TODO:
# user master 데이터에서 고유 사용자 수를 계산하세요.
# 또한 country와 device의 분포를 value_counts()로 출력하여
# 사용자 구성이 어떻게 이루어져 있는지 확인하세요.
print("user 수:", user_df['user_id'].nunique())
print(user_df['country'].value_counts())
print(user_df['device'].value_counts())
```

```python
# TODO:
# user master 데이터에서 고유 사용자 수를 계산하세요.
# 또한 country와 device의 분포를 value_counts()로 출력하여
# 사용자 구성이 어떻게 이루어져 있는지 확인하세요.
print("user 수:", user_df['user_id'].nunique())
print(user_df['country'].value_counts())
print(user_df['device'].value_counts())
```

```python
# TODO:
# exposure log에서 총 노출 수를 계산하세요.
# 그리고 user_id 기준으로 그룹화하여 사용자별 노출 수를 구한 뒤,
# 그 평균을 계산하세요.
# 마지막으로 variant 분포를 비율 기준(normalize=True)으로 확인하세요.
print("총 노출 수:", len(exposure_df))

user_exposure = exposure_df.groupby('user_id').size()
print("평균 노출 수:", user_exposure.mean())

print(exposure_df['variant'].value_counts(normalize=True))
```

```python
# exposure_df의 user_id가 모두 user_df의 user_id 안에 포함되는지 확인하세요.
# 또한 action_df의 user_id가 exposure_df의 user_id 안에 포함되는지도 확인하세요.
# 집합(set)과 issubset()을 사용하면 포함 관계를 점검할 수 있습니다.
print("exposure 데이터의 user_id가 user_master에 모두 포함되어 있는가:",
      set(exposure_df["user_id"]).issubset(set(user_df["user_id"])))

print("action 데이터의 user_id가 exposure 데이터에 모두 포함되어 있는가:",
      set(action_df["user_id"]).issubset(set(exposure_df["user_id"])))
```
