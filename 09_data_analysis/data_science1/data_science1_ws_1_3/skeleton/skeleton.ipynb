```python
import pandas as pd
```

```python
# lv2에서 생성한 데이터를 사용합니다.
df = pd.read_csv("../data/ab_user_level_final.csv")
```

```python
# TODO:
# variant가 0인 사용자를 group_A,
# variant가 1인 사용자를 group_B로 분리하세요.
group_A = df[df['variant'] == 0]
group_B = df[df['variant'] == 1]
```

```python
# TODO:
# CTR은 click이 1번 이상 발생한 사용자 비율입니다.
# 각 그룹에서 (click > 0)의 평균을 계산하세요.
ctr_A = (group_A['click'] > 0).mean()
ctr_B = (group_B['click'] > 0).mean()

# TODO:
# Conversion은 purchase가 1번 이상 발생한 사용자 비율입니다.
# 각 그룹에서 (purchase > 0)의 평균을 계산하세요.
conv_A = (group_A['purchase'] > 0).mean()
conv_B = (group_B['purchase'] > 0).mean()

# TODO:
# Revenue는 사용자당 평균 매출입니다.
# 각 그룹에서 revenue 평균을 계산하세요.
rev_A = group_A['revenue'].mean()
rev_B = group_B['revenue'].mean()
```

```python
# TODO:
# metric, A, B 컬럼을 가지는 비교용 DataFrame을 생성하세요.
# metric에는 CTR, Conversion, Revenue가 들어가야 합니다.
result = pd.DataFrame({
    "metric": ["CTR", "Conversion", "Revenue"],
    "A": [ctr_A, conv_A,rev_A],
    "B": [ctr_B, conv_B,rev_B]
})

result
```

```python
# TODO:
# B 그룹 값에서 A 그룹 값을 뺀 diff (B-A) 컬럼을 추가하세요.
result["diff (B-A)"] = result["B"] - result["A"]

print(result)
```

```python
print("\n=== 해석 ===")

for _, row in result.iterrows():
    metric = row["metric"]
    diff = row["diff (B-A)"]

    if diff > 0:
        print(f"{metric}: B 그룹이 A 그룹보다 높습니다. (차이: {diff:.4f})")
    elif diff < 0:
        print(f"{metric}: A 그룹이 B 그룹보다 높습니다. (차이: {abs(diff):.4f})")
    else:
        print(f"{metric}: 두 그룹 간 차이가 없습니다.")
```
