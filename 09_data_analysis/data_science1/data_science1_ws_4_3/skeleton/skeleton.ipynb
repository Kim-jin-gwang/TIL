```python
"""
실시간(준실시간) 날씨 Polling 실습

- OpenAPI(REST)로 현재 날씨를 주기적으로 조회
- Open-Meteo API 사용
- 핵심 포인트: 요청 -> 응답 -> 대기 -> 다시 요청
"""

import time
import requests
```

```python
def fetch_weather(lat: float, lon: float) -> dict:
    """
    현재 날씨를 OpenAPI로 1회 조회하는 함수

    매개변수
    - lat: 위도
    - lon: 경도

    반환값
    - current 영역의 딕셔너리 데이터
      예:
      {
          "time": "2026-04-02T11:30",
          "temperature_2m": 14.2,
          "weather_code": 3
      }
    """

    # Open-Meteo 현재 날씨 조회용 API 주소
    url = "https://api.open-meteo.com/v1/forecast"

    # API에 전달할 요청 파라미터
    # latitude, longitude: 조회할 위치 정보
    # current: 현재 시점에 필요한 날씨 항목 지정
    params = {
        "latitude": lat,
        "longitude": lon,
        "current": "temperature_2m,weather_code",
    }

    # TODO:
    # requests.get()을 사용해서 API를 호출하세요.
    # 조건:
    # - url 전달
    # - params 전달
    # - timeout=10 설정
    resp = requests.get(url=url, params=params, timeout=10)

    # TODO:
    # 응답 상태 코드가 비정상(예: 404, 500 등)일 경우
    # 예외를 발생시키도록 작성하세요.
    resp.raise_for_status()

    # TODO:
    # JSON 응답에서 "current" 영역만 추출해서 반환하세요.
    return resp.json()["current"]
```

```python
def poll_weather(lat: float, lon: float, interval_sec: int = 60):
    """
    일정 시간 간격으로 현재 날씨를 반복 조회하는 함수

    매개변수
    - lat: 위도
    - lon: 경도
    - interval_sec: 몇 초마다 다시 조회할지 설정하는 간격

    동작 방식
    1. 현재 날씨 조회
    2. 결과 출력
    3. interval_sec초 대기
    4. 다시 조회
    """

    # TODO:
    # Polling은 보통 계속 반복해서 데이터를 수집하므로
    # 무한 반복 구조로 작성하세요.
    while True:
        try:
            # TODO:
            # fetch_weather()를 호출해서 현재 날씨 데이터를 가져오세요.
            current = fetch_weather(lat,lon)

            # current 딕셔너리에서 필요한 값을 꺼내 출력
            # - time: 측정 시각
            # - temperature_2m: 기온
            # - weather_code: 날씨 코드
            print(
                f"[{current['time']}] "
                f"temp={current['temperature_2m']}°C "
                f"code={current['weather_code']}"
            )

        except Exception as e:
            # API 호출 실패, 네트워크 오류, 응답 형식 문제 등이 발생하면
            # 에러 메시지를 출력하고 다음 반복으로 넘어감
            print("API 호출 실패:", e)

        # interval_sec초 동안 대기한 뒤 다시 요청하세요.
        time.sleep(interval_sec)
```

```python
# 조회할 위치의 위도/경도
LAT = 36.3553
LON = 127.2983

# TODO:
# 30초마다 현재 날씨를 조회하도록 poll_weather()를 실행하세요.
poll_weather(LAT,LON,30)
```
