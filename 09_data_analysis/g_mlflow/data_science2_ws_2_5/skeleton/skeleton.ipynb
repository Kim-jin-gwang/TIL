```python
import mlflow.sklearn
import pandas as pd
```

```python
mlflow.set_tracking_uri("http://127.0.0.1:5000")
```

```python
# 1. 저장된 모델의 run_id 입력
run_id = "b0c993f79b24406eb3251c99a4944edf" # 예: "a9273033ec4a4b3ba5e9587f49383c66"
logged_model_uri = f"runs:/{run_id}/model"
```

```python
# 2. 모델 로드
model = mlflow.sklearn.load_model(logged_model_uri)
```

```python
# 3. 새로운 데이터 불러오기
new_data = pd.read_csv("../data/new_credit_transactions.csv")
```

```python
# 4. 예측 수행
predictions = model.predict(new_data)
```

```python
# 5. 결과 저장
new_data["predictions"] = predictions
new_data.to_csv("../data/predictions.csv", index=False)

print("예측 완료! 'data/predictions.csv' 파일로 저장됨")
```
