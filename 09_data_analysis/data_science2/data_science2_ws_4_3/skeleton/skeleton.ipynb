```python
# 시나리오 설명
'''
김싸피는 판매 데이터를 분석하여 매출 기준으로 고객 등급을 분류하려 합니다.
quantity * unitprice를 활용해 거래당 수익(total_revenue)을 계산하고,
특정 기준 이상인 고객을 "VIP"로 태깅하여 이후 분석 대상으로 따로 분리합니다.
또한 날짜 정보를 유지해 향후 시계열 분석에도 활용할 수 있도록 합니다.
'''
```

```python
# 요구 사항
'''
filtered_sales.csv 파일을 불러옵니다.
quantity * unitprice로 total_revenue 파생 컬럼을 생성합니다.
수익이 1000 이상이면 "VIP", 아니면 "Normal"로 분류해 grade 컬럼에 저장합니다.
"VIP" 고객만 필터링하여 저장합니다.
결과는 output/vip_sales.csv로 저장합니다.
'''
```

```python
import pandas as pd
import os
```

```python
def add_grade_column_and_filter_vip(df, revenue_threshold=1000):
    """
    매출 기준으로 고객 등급을 분류하고, VIP만 필터링
    """

    # 0. 총 매출액 계산: 수량(quantity) * 단가(unitprice)
    df["total_revenue"] = df["quantity"] * df["unitprice"]

    # 1. 결측치 제거: 매출 정보가 없는 행은 제외
    df = df.dropna(subset=["total_revenue"])

    # 2. 고객 등급 지정: 기준치 이상이면 VIP, 아니면 Normal
    df["grade"] = df["total_revenue"].apply(
        lambda x: "VIP" if x >= revenue_threshold else "Normal"
    )

    # 3. VIP 등급 고객만 필터링
    vip_df = df[df['grade'] == 'VIP']

    return vip_df
```

```python
# 경로
input_path = "../data/filtered_sales.csv"
output_path = "../output/vip_sales.csv"

output_dir = os.path.dirname(output_path)
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    print(f"폴더가 존재하지 않아 생성했습니다: {output_dir}")
```

```python
# 로딩
df = pd.read_csv(input_path)
```

```python
# VIP 필터링
vip_df = add_grade_column_and_filter_vip(df)
```

```python
# 저장
vip_df.to_csv(output_path, index=False)
```
