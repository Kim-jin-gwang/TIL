```python
# 시나리오 설명
'''
김싸피는 새로운 전자상거래 대용량 데이터를 Kaggle에서 수집했습니다.
이 파일은 100,000건 이상의 주문 정보를 담고 있으며, 다양한 부서에서 활용할 수 있도록
분석에 필요한 컬럼만 추출하고 열 이름을 정제한 뒤 저장하려고 합니다.
사용 데이터셋 : Online sales dataset
https://www.kaggle.com/datasets/yusufdelikkaya/online-sales-dataset
'''
```

```python
# 요구 사항
'''
Kaggle에서 받은 online_sales_dataset.csv 파일을 청크 단위로 불러옵니다.
아래 컬럼들만 선택해 추출합니다:
InvoiceNo, Description, Quantity, UnitPrice, Discount, ShippingCost, Category, SalesChannel, ReturnStatus, OrderPriority
열 이름을 소문자 + 공백 및 특수문자를 언더스코어(_)로 정제합니다.
병합된 최종 결과를 output/cleaned_sales.csv로 저장하세요.
'''
```

```python
import pandas as pd
import os
import re
```

```python
# 대용량 데이터를 청크 단위로 불러오고 정제하는 함수 작성
def load_and_standardize_chunks(file_path, use_cols, chunksize=10000):
    """
    주어진 컬럼만 선택해 청크 단위로 데이터를 로딩하고,
    컬럼명을 소문자 + snake_case로 정제한다.
    """
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"파일이 없습니다: {file_path}")
    
    dfs = []
    for chunk in pd.read_csv(file_path, usecols=use_cols, chunksize=chunksize, encoding="ISO-8859-1"):
        # 열 이름 정제: 소문자로 변환하고 특수문자 제거
        new_cols = []
        for col in chunk.columns:
            clean = col.lower()  # ← 열 이름을 소문자로 변환하세요
            clean = re.sub(r'[^a-z0-9]', '_', clean)
            new_cols.append(clean)
        chunk.columns = new_cols
        
        dfs.append(chunk)
    
    df = pd.concat(dfs, )  # ← 리스트의 모든 DataFrame을 병합하세요
    return df
```

```python
# 필요한 컬럼 정의
# 사용할 컬럼 목록 정의
# 데이터셋을 먼저 전체 불러보고 주요 컬럼을 선정합니다.
# 예: ["InvoiceNo", "StockCode", "Description", "Quantity", "InvoiceDate", "UnitPrice", "CustomerID", "Country
use_cols = [
    "InvoiceNo", "Description", "Quantity", "InvoiceDate", "UnitPrice",
    "Discount", "ShippingCost", "Category",
    "SalesChannel", "ReturnStatus", "OrderPriority"
]
```

```python

# 경로 설정
input_path = "../data/online_sales_dataset.csv"
output_path = "../output/cleaned_sales.csv"
```

```python
# 데이터 로딩 및 정제
df = load_and_standardize_chunks(input_path, use_cols)
```

```python
# 저장
os.makedirs("../output", exist_ok=True)
df.to_csv(output_path, index=False)
```
