```python
# 시나리오 설명
'''
데이터를 분석 가능한 상태로 만들기 위해 김싸피는 결측치와 데이터 타입을 정제하려고 합니다.
컬럼마다 결측치 비율을 분석해 기준 이상이면 삭제하고, object 타입 중 숫자로 변환 가능한 값들은
수치형(float)으로 바꿔 향후 분석 및 시각화에 활용할 수 있도록 합니다.
'''
```

```python
# 요구 사항
'''
practice1/output/cleaned_sales.csv 파일을 불러옵니다.
컬럼별 결측치 비율을 계산하고 출력합니다.
결측치 비율이 40% 이상인 컬럼은 제거합니다.
object 타입 중 숫자형으로 변환 가능한 컬럼은 float으로 변환합니다.
최종 정제된 데이터는 output/filtered_sales.csv에 저장합니다.
'''
```

```python
import pandas as pd
```

```python
def clean_missing_and_convert_types(df, null_threshold=0.4):
    """
    결측치 기준 이상인 컬럼 제거 + object → float 변환
    """
    # 1. 컬럼별 결측치 비율 계산
    null_ratio = df.isnull().mean().round(3)  # ← 결측치 비율 계산 (소수점 3자리 반올림)
    print("컬럼별 결측치 비율:")
    print(null_ratio)

    # 2. 기준 이상 컬럼 제거
    drop_cols = null_ratio[null_ratio >= null_threshold].index.tolist()  # ← 비율이 threshold 이상인 컬럼 이름 추출 후 리스트로 변환
    df = df.drop(columns = drop_cols)         # ← 해당 컬럼들 제거

    # 3. object 타입 → 수치형(float) 변환
    for col in df.select_dtypes(include="object").columns:
        try:
            converted = pd.to_numeric(df[col], errors="coerce")
            if converted.notnull().sum() > 0:
                df[col] = converted
        except Exception as e:
            print(f"{col} 변환 실패:", e)

    # 4. 결측치 개수 요약
    null_summary = df.isnull().sum().to_dict()
    print("결측치 개수 요약:")
    print(null_summary)

    return df
```

```python
import os

# 경로 설정
input_path = "cleaned_sales.csv"
output_path = "../output/filtered_sales.csv"

output_dir = os.path.dirname(output_path)
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    print(f"폴더가 존재하지 않아 생성했습니다: {output_dir}")
```

```python
# 로딩
df = pd.read_csv(input_path)
```

```python
# 정제
df_cleaned = clean_missing_and_convert_types(df)
```

```python
# 저장
df_cleaned.to_csv(output_path, index=False)
```
