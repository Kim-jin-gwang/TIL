```python
# 시나리오 설명
'''
김싸피는 VIP 고객 데이터를 분석 부서별로 분배하려고 합니다.
특히 제품 카테고리(category)별로 구분해 각 팀에 전달하려 하며,
각 그룹은 개별 CSV 파일로 저장되어야 합니다.
또한 저장된 결과를 간단히 확인하여 제대로 나뉘었는지도 검증합니다.
'''
```

```python
# 요구 사항
'''
vip_sales.csv 파일을 불러옵니다.
category 컬럼을 기준으로 그룹핑합니다.
각 카테고리별로 vip_<카테고리>.csv 형식으로 저장하세요.
카테고리명은 공백 → 언더스코어, 모두 소문자 변환
저장된 각 파일의 이름과 상위 2개 행을 출력하여 확인합니다.
결과는 output/ 폴더에 저장됩니다.
'''
```

```python
import pandas as pd
import os
```

```python
def save_by_category(df, category_col="category", prefix="vip"):
    """
    특정 컬럼을 기준으로 데이터를 그룹화하여,
    각 그룹을 개별 파일로 저장합니다.
    """
    # 1. 그룹 나누기: 지정된 컬럼(category_col)을 기준으로 묶기
    grouped = df.groupby(category_col)  # ← 어떤 메서드로 그룹화할 수 있을지 생각해보세요

    # 2. 출력 폴더 생성
    os.makedirs("../output", exist_ok=True)
    saved_files = []

    for cat, group in grouped:
        # 3. 카테고리명을 파일명에 적합한 문자열로 변환
        #    예: "Office Supplies" → "office_supplies"
        cat_safe = cat.strip().lower().replace(" ", "_")

        # 4. 파일 경로 지정
        file_path = f"../output/{prefix}_{cat_safe}.csv"

        # 5. 그룹 데이터를 CSV로 저장
        group.to_csv(file_path, index=False)

        saved_files.append(file_path)

    print(f"총 {len(saved_files)}개 파일이 저장되었습니다.")
    return saved_files
```

```python
# 입력 경로
input_path = "../data/vip_sales.csv"
df = pd.read_csv(input_path)
```

```python
# 저장
saved_files = save_by_category(df)
```

```python
# 결과 확인
print("\n저장된 파일 목록 및 샘플 출력:")
for path in saved_files:
    print(f"\n[파일] {os.path.basename(path)}")
    try:
        preview_df = pd.read_csv(path)
        print(preview_df.head(2))
    except Exception as e:
        print(f"읽기 실패: {e}")
```
