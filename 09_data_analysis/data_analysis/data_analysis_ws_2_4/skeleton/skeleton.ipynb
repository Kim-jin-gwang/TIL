```python
import pandas as pd
import seaborn as sns

# Seaborn에서 타이타닉 데이터셋 로드
titanic = sns.load_dataset('titanic')

# 데이터셋 확인
print(titanic.head())
```

```python
# 1등석 승객만 선택
first_class_sample = titanic[titanic['class']=='First']  # TODO: 1등석 승객만 선택

# 1등석 승객에서 생존 확률 계산
first_class_survival_prob = first_class_sample['survived'].mean()  # TODO: 1등석 승객의 생존 확률 계산

# 출력
print(f"1등석 승객 생존 확률: {first_class_survival_prob:.2f}")
```

```python
import numpy as np

# 'age' 컬럼에서 임의로 표본 100개 추출
random_sample = titanic.sample(n=100, random_state=42)

# 임의 추출된 표본에서 생존 확률 계산
random_sample_survival_prob = random_sample['survived'].mean()  # TODO: 임의 추출된 표본의 생존 확률 계산

# 출력
print(f"임의 추출된 표본의 생존 확률: {random_sample_survival_prob:.2f}")
```

```python
# 각 pclass별 10% 샘플링 (층화 샘플링)
stratified_sample = (
    titanic
    .groupby("pclass")
    .sample(frac=0.1, random_state=42)
)
# 층화 샘플에서 생존 확률 계산
stratified_sample_survival_prob = stratified_sample['survived'].mean()  # TODO: 층화 샘플링을 통한 생존 확률 계산

# 출력
print(f"층화 샘플링을 통한 생존 확률: {stratified_sample_survival_prob:.2f}")
```
