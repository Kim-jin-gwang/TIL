```python
#  1. 라이브러리 불러오기 
import pandas as pd
```

```python
#  2. 데이터 로드 
# 판매 데이터 파일 경로 설정
sales_file_path = "../data/sales_data.csv"
# CSV 파일을 데이터프레임으로 로드
df_sales = pd.read_csv(sales_file_path)  # pandas를 사용하여 CSV 파일을 로드
```

```python
#  3. 데이터 확인 
# 데이터프레임 기본 정보 출력
print("=== 판매 데이터 정보 ===")
df_sales.info()  # 데이터프레임의 컬럼 정보 및 데이터 타입 확인

# 데이터의 상위 5개 행 출력
print("\n=== 판매 데이터 샘플 ===")
df_sales.info()
```

```python
#  4. 결측치 및 이상치 처리 
print("\n=== 결측치 개수 확인 ===")
df_sales.isna().sum()  # 각 컬럼의 결측치 개수를 출력

print("\n=== 기본 통계 정보 확인 ===")
df_sales.describe() # 데이터의 평균, 표준편차, 최소/최대값 확인
```

```python
#  5. 매출(TotalRevenue) 컬럼 생성 
df_sales["TotalRevenue"] = df_sales['Price'] * df_sales['SalesVolume']  # 가격과 판매량을 곱하여 매출을 계산

print("\n=== 매출(TotalRevenue) 컬럼 추가 후 데이터 확인 ===")
df_sales.head()
```

```python
#  6. 주요 변수 간 상관관계 분석 
correlation_matrix = df_sales[["Price", "SalesVolume", "DiscountRate", "AdvertisingExpense", "TotalRevenue"]].corr()

print("\n=== 주요 변수 간 상관관계 분석 ===")
correlation_matrix
```

```python
#  7. 데이터 분석 
# 1) 제품별 평균 판매량 분석
df_avg_sales = df_sales.groupby('ProductID')["SalesVolume"].mean().reset_index()  # 그룹화할 컬럼과 평균을 낼 컬럼 입력
df_avg_sales = df_avg_sales.sort_values(by="SalesVolume", ascending=False)  # 정렬할 기준 컬럼 입력

print("\n=== 제품별 평균 판매량 상위 10개 ===")
df_avg_sales.head(10)
```

```python
# 2) 할인율이 높은 제품과 낮은 제품 비교
df_high_discount = df_sales.nlargest(10, "DiscountRate")[["ProductID", "Price", "SalesVolume", "AdvertisingExpense"]]
df_low_discount = df_sales.nsmallest(10, "DiscountRate")[["ProductID", "Price", "SalesVolume", "AdvertisingExpense"]]

print("\n=== 할인율이 높은 제품 10개 ===")
print(df_high_discount)

print("\n=== 할인율이 낮은 제품 10개 ===")
print(df_low_discount)
```

```python
# 3) 광고비가 높은 제품과 낮은 제품 비교
df_high_ad = df_sales.nlargest(10, "AdvertisingExpense")[["ProductID", "Price", "SalesVolume", "AdvertisingExpense"]]
df_low_ad = df_sales.nsmallest(10, "AdvertisingExpense")[["ProductID", "Price", "SalesVolume", "AdvertisingExpense"]]

print("\n=== 광고비가 높은 제품 10개 ===")
print(df_high_ad)

print("\n=== 광고비가 낮은 제품 10개 ===")
print(df_low_ad)
```

```python
# 4) 매출 상위 10개 제품 분석
df_top_revenue = df_sales.nlargest(10, "TotalRevenue")[["ProductID", "Price", "SalesVolume", "TotalRevenue"]]

print("\n=== 매출 상위 10개 제품 ===")
print(df_top_revenue)

# 위와 동일한 내용을 sort_values를 사용하여 구현
df_top_revenue = df_sales.sort_values(by="TotalRevenue",ascending=False).head(10)[["ProductID", "Price", "SalesVolume", "TotalRevenue"]]

print("\n=== 매출 상위 10개 제품 ===")
print(df_top_revenue)
```
