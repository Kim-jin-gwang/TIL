```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'Malgun Gothic'

# 샘플 데이터
scores = [82, 92, 88, 79, 85, 91, 94, 77, 82, 90]

# DataFrame으로 변환
df = pd.DataFrame(scores, columns=['Scores'])
```

```python
# 기초 통계량 계산
mean_score = df['Scores'].mean()  # TODO: 평균 계산
median_score = df['Scores'].median()  # TODO: 중앙값 계산
mode_score = df['Scores'].mode()[0]  # TODO: 최빈값 계산
std_dev = df['Scores'].std()  # TODO: 표준편차 계산
iqr = df['Scores'].quantile(0.75) - df['Scores'].quantile(0.25)  # TODO: IQR 계산
```

```python
# 결과 출력
print(f'평균: {mean_score}, 중앙값: {median_score}, 최빈값: {mode_score}')
print(f'표준편차: {std_dev}, IQR: {iqr}')
```

```python
# 히스토그램
df['Scores'].plot(kind='hist', bins=5, alpha=0.7, color='blue', edgecolor='black')
plt.title('점수 분포')
plt.xlabel('점수')
plt.show()
```
