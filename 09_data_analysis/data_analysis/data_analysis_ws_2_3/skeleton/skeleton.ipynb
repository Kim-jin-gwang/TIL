```python
# 필요한 패키지 임포트
import numpy as np
import pandas as pd

# 모집단 정의: 성별에 대한 500명 남성, 500명 여성 데이터 생성
population = ['Male'] * 500 + ['Female'] * 500

# 무작위 표본 추출: 모집단에서 무작위로 100명 샘플링 (replace=False는 중복 없이 추출)
random_sample = np.random.choice(population, 100, replace=False)

# 각 데이터에 대해 선택 확률 부여
# Male은 높은 확률, Female은 낮은 확률
prob = [0.7 if gender == 'Male' else 0.3 for gender in population]

# 확률 정규화 (합이 1이 되도록)
prob = np.array(prob) / np.sum(prob)

biased_sample = np.random.choice(population,100,replace=False,p=prob)

# 표본에서 성별 분포 계산
# 무작위 표본의 성별 분포 계산
random_dist = pd.Series(random_sample).value_counts()  # TODO: 성별 분포 계산

# 편향된 표본의 성별 분포 계산
biased_dist = pd.Series(biased_sample).value_counts()  # TODO: 성별 분포 계산

# 출력: 무작위 표본과 편향된 표본의 성별 분포 확인
print('무작위 표본 분포:')
print(random_dist)
print('\n편향된 표본 분포:')
print(biased_dist)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정
plt.rcParams['font.family'] = 'Malgun Gothic'

fig, ax = plt.subplots(1, 2, figsize=(12, 6))

sns.barplot(x=random_dist.index, y=random_dist.values, ax=ax[0], palette='Blues', hue=random_dist.index)
ax[0].set_title('무작위 표본 분포')

sns.barplot(x=biased_dist.index, y=biased_dist.values, ax=ax[1], palette='Reds', hue=biased_dist.index)
ax[1].set_title('편향된 표본 분포')

plt.tight_layout()
plt.show()
```
