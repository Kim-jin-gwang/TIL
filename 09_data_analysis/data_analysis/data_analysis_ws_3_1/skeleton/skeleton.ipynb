```python
import pandas as pd
import seaborn as sns

# Seaborn에서 타이타닉 데이터셋 로드
titanic = sns.load_dataset('titanic')

# 데이터셋 확인
print(titanic.head())
```

```python
# 성별에 따른 생존자 비율 계산
survival_by_gender = titanic.groupby('sex')['survived'].mean()

# 출력
print("성별에 따른 생존 확률:")
print(survival_by_gender)
```

```python
# 전체 승객 중 생존한 승객 비율 (P(Survived = 1))
p_survived = titanic['survived'].mean()  # TODO: 전체 승객 중 생존한 승객 비율 계산
```

```python
# 전체 승객 중 여성 승객 비율 (P(Female))
p_female = (titanic['sex']=='female').mean()  # TODO: 전체 여성 승객 비율 계산
```

```python
# 여성 승객 중 생존한 승객 비율 (P(Survived = 1 | Female))
p_female_survived = titanic[titanic['sex']=='female']['survived'].mean()  # TODO: 여성 승객 중 생존한 승객 비율 계산
```

```python
# P(Female | Survived = 1) - 베이즈 정리 적용
# P(Female | Survived = 1) = (P(Survived = 1 | Female) * P(Female)) / P(Survived = 1)
p_female_given_survived = (p_female_survived * p_female) / p_survived
```

```python
# 출력
print(f"여성 승객 중 생존한 확률 (P(Survived = 1 | Female)): {p_female_survived:.2f}")
print(f"생존한 승객 중 여성 승객일 확률 (P(Female | Survived = 1)): {p_female_given_survived:.2f}")
```
